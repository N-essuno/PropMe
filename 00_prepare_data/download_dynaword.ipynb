Load alpaca dataset and save it as JSONL file where instruction, input and output are converted to text fields and are saved separately.

```python
from datasets import load_dataset
import json
```

```python
dynaword = load_dataset("danish-foundation-models/danish-dynaword")
dynaword
```

```python
dynaword_train = dynaword['train']
dynaword_train
```

```python
# Create a json string for each row in the dataset with fields "instruction", "input" and "output" not empty
json_strings = []
for id, row in enumerate(dynaword_train):
    text = row['text']
    id = row['id']
    source = row['source']
    token_count = row['token_count']
    added = row['added']
    created = row['created']

    if text != "":
        json_strings.append({
            "id": id,
            "text": text,
            "source": source,
            "token_count": token_count,
            "added": added,
            "created": created
        })
```

```python
jsonl_path = "../00_data/dynaword/dynaword.jsonl"
```

```python
with open(jsonl_path, "w", encoding="utf-8", newline="\n") as f:
    for obj in json_strings:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")
```
